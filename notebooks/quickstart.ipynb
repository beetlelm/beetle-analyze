# beetle-analyze quickstart

Install → run the headline analysis on the bundled toy data → read the number → change one variable → re-run.

Everything here runs on CPU in under a second with no model downloads.

## 1. Install

From a clone of the repo:

```bash
pip install -e .
```

(or `pip install beetle-analyze` once it is on PyPI).

````python
from beetle_analyze import analyze_file, toy_data_path

# The bundled synthetic eval output: one BLiMP-style minimal pair per line,
# each with the model's total log-prob for the good and bad sentence.
path = toy_data_path()
print(path)
````

## 2. Run the headline analysis

The headline metric is BLiMP accuracy: the fraction of minimal pairs where the
grammatical sentence gets a higher score than the ungrammatical one.

````python
result = analyze_file(path, scoring="sum")
print("overall accuracy:", result.overall_accuracy)   # expected: 0.60
print("95% CI:", (result.ci_low, result.ci_high))
result.per_phenomenon
````

## 3. Change one variable and re-run

Raw-`sum` scoring has a length bias (longer sentences score lower just for being
longer). Switch to per-token `mean` scoring to remove it. Watch the accuracy move.

````python
mean_result = analyze_file(path, scoring="mean")
print("sum  scoring:", result.overall_accuracy)        # 0.60
print("mean scoring:", mean_result.overall_accuracy)   # 0.80
mean_result.per_phenomenon
````

## 4. Bring your own eval output

Point `analyze_file` at any JSONL with the fields `phenomenon`, `logprob_good`,
`logprob_bad` (and optionally `n_tokens_good` / `n_tokens_bad` for mean scoring).
You can produce such a file from your own model by summing per-token log-probs
over each sentence in a minimal-pair benchmark.
